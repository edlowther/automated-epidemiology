```python
import json
```

```python
for topic in ['bw_bmi', 'ses_mental_health', 'maternal_depression']:
    with open(f'{topic}.json', 'r') as input, open(f'{topic}.md', 'a') as output:
        topic_json = json.load(input)
        output.write('\n')
        for idea in topic_json:
            output.write(f"## {idea['Title']}\n")
            output.write("### Short Hypothesis\n")
            output.write(f'{idea["Short Hypothesis"]}\n')
            output.write("### Related Work\n")
            output.write(f'{idea["Related Work"]}\n')
            output.write("### Abstract\n")
            output.write(f'{idea["Abstract"]}\n')
            output.write("### Experiments\n")
            for experiment in idea["Experiments"]:
                output.write(f"- {experiment}\n")
            output.write("### Risk Factors and Limitations\n")
            for limitation in idea["Risk Factors and Limitations"]:
                output.write(f"- {limitation}\n")
```
